```python
import pandas as pd
import re
import html
from urllib.parse import unquote

# =========================
# 1. CSVファイルの読み込み
# =========================

train_path = "/content/drive/MyDrive/Kaggle/NLP/train.csv"
test_path = "/content/drive/MyDrive/Kaggle/NLP/test.csv"

train = pd.read_csv(train_path)
test = pd.read_csv(test_path)

print("読み込み完了")
print("train shape:", train.shape)
print("test shape :", test.shape)


# =========================
# 2. 必要カラムの確認
# =========================

required_train_cols = ["id", "keyword", "location", "text", "target"]
required_test_cols = ["id", "keyword", "location", "text"]

missing_train_cols = [col for col in required_train_cols if col not in train.columns]
missing_test_cols = [col for col in required_test_cols if col not in test.columns]

if missing_train_cols:
    raise ValueError(f"train.csvに存在しないカラムがあります: {missing_train_cols}")

if missing_test_cols:
    raise ValueError(f"test.csvに存在しないカラムがあります: {missing_test_cols}")


# =========================
# 3. 軽いテキストクリーニング関数
# =========================

def clean_text_for_tfidf(text):
    """
    TF-IDF + LogisticRegression向けの軽いクリーニング。
    過剰な正規化はせず、単語情報をできるだけ残す。
    """
    if pd.isna(text):
        return ""

    text = str(text)

    # HTMLエスケープを戻す
    text = html.unescape(text)

    # URLエンコードを戻す
    # 例: body%20bags -> body bags
    text = unquote(text)

    # 小文字化
    text = text.lower()

    # URLを <url> に置換
    text = re.sub(r"https?://\S+|www\.\S+", " <url> ", text)

    # メンションを <user> に置換
    text = re.sub(r"@\w+", " <user> ", text)

    # ハッシュタグは # だけ削除して単語は残す
    text = re.sub(r"#", "", text)

    # 連続する空白を1つにまとめる
    text = re.sub(r"\s+", " ", text)

    # 前後の空白を削除
    text = text.strip()

    return text


# =========================
# 4. train / test に同じ前処理を適用
# =========================

train_processed = train.copy()
test_processed = test.copy()

# keywordの欠損値は空文字で補完
train_processed["keyword"] = train_processed["keyword"].fillna("")
test_processed["keyword"] = test_processed["keyword"].fillna("")

# textにも念のため欠損対応
train_processed["text"] = train_processed["text"].fillna("")
test_processed["text"] = test_processed["text"].fillna("")

# keywordとtextをクリーニング
train_processed["keyword_clean"] = train_processed["keyword"].apply(clean_text_for_tfidf)
test_processed["keyword_clean"] = test_processed["keyword"].apply(clean_text_for_tfidf)

train_processed["text_clean"] = train_processed["text"].apply(clean_text_for_tfidf)
test_processed["text_clean"] = test_processed["text"].apply(clean_text_for_tfidf)

# keyword_clean と text_clean を結合して input_text を作成
train_processed["input_text"] = (
    train_processed["keyword_clean"] + " " + train_processed["text_clean"]
).str.strip()

test_processed["input_text"] = (
    test_processed["keyword_clean"] + " " + test_processed["text_clean"]
).str.strip()


# =========================
# 5. 確認
# =========================

print("\n===== train_processed head =====")
display(train_processed.head())

print("\n===== test_processed head =====")
display(test_processed.head())

print("\n===== train_processed missing values =====")
display(train_processed.isnull().sum().sort_values(ascending=False))

print("\n===== test_processed missing values =====")
display(test_processed.isnull().sum().sort_values(ascending=False))

print("\n===== input_text samples: train =====")
display(train_processed[["id", "keyword", "keyword_clean", "text", "text_clean", "input_text", "target"]].head(10))

print("\n===== input_text samples: test =====")
display(test_processed[["id", "keyword", "keyword_clean", "text", "text_clean", "input_text"]].head(10))

print("\n===== target distribution =====")
display(train_processed["target"].value_counts())
display(train_processed["target"].value_counts(normalize=True))

print("\n前処理完了")
print("train_processed shape:", train_processed.shape)
print("test_processed shape :", test_processed.shape)
```

```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.pipeline import Pipeline
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression

from sklearn.metrics import (
    accuracy_score,
    f1_score,
    classification_report,
    confusion_matrix
)

# =========================
# 1. 使用するデータの準備
# =========================

# 前処理済みの input_text を使用
X = train_processed["input_text"].fillna("")
y = train_processed["target"].astype(int)

X_test = test_processed["input_text"].fillna("")

print("X shape:", X.shape)
print("y shape:", y.shape)
print("X_test shape:", X_test.shape)

print("\n===== target distribution =====")
display(y.value_counts())
display(y.value_counts(normalize=True))


# =========================
# 2. 学習用データ / 検証用データに分割
# =========================

X_train, X_valid, y_train, y_valid = train_test_split(
    X,
    y,
    test_size=0.2,
    stratify=y,
    random_state=42
)

print("\n===== split result =====")
print("X_train:", X_train.shape)
print("X_valid:", X_valid.shape)
print("y_train:", y_train.shape)
print("y_valid:", y_valid.shape)


# =========================
# 3. TF-IDF + LogisticRegression Pipeline
# =========================

model = Pipeline([
    (
        "tfidf",
        TfidfVectorizer(
            ngram_range=(1, 2),
            min_df=2,
            max_df=0.95,
            sublinear_tf=True
        )
    ),
    (
        "logreg",
        LogisticRegression(
            max_iter=1000,
            C=1.0,
            random_state=42
        )
    )
])


# =========================
# 4. モデル学習
# =========================

model.fit(X_train, y_train)

print("\n学習完了")


# =========================
# 5. 検証データで予測
# =========================

valid_pred = model.predict(X_valid)

accuracy = accuracy_score(y_valid, valid_pred)
f1 = f1_score(y_valid, valid_pred)

print("\n===== validation score =====")
print("accuracy:", accuracy)
print("f1_score:", f1)

print("\n===== classification report =====")
print(classification_report(y_valid, valid_pred))

print("\n===== confusion matrix =====")
cm = confusion_matrix(y_valid, valid_pred)
cm_df = pd.DataFrame(
    cm,
    index=["actual_0", "actual_1"],
    columns=["pred_0", "pred_1"]
)
display(cm_df)


# =========================
# 6. testデータに対して予測
# =========================

test_pred = model.predict(X_test)

print("\n===== test prediction distribution =====")
display(pd.Series(test_pred).value_counts())
display(pd.Series(test_pred).value_counts(normalize=True))


# =========================
# 7. Kaggle提出用CSVを作成
# =========================

submission = pd.DataFrame({
    "id": test_processed["id"],
    "target": test_pred.astype(int)
})

submission_path = "submission_tfidf_logreg.csv"
submission.to_csv(submission_path, index=False)

print("\n提出用CSVを作成しました:", submission_path)

print("\n===== submission head =====")
display(submission.head())

print("\n===== submission shape =====")
print(submission.shape)


# =========================
# 8. 改善候補メモ
# =========================

# 改善候補1:
# char-level TF-IDFを追加する
# 例: analyzer="char_wb", ngram_range=(3, 5)
#
# 改善候補2:
# LogisticRegressionのCを調整する
# 例: C=0.5, 1.0, 2.0, 3.0 など
#
# 改善候補3:
# class_weight="balanced" を試す
#
# 改善候補4:
# word TF-IDF と char TF-IDF を FeatureUnion で結合する
```

```python
import pandas as pd
import numpy as np
import re
import html
from urllib.parse import unquote

# =========================
# 1. CSVファイルの読み込み
# =========================

train_path = "/content/drive/MyDrive/Kaggle/NLP/train.csv"
test_path = "/content/drive/MyDrive/Kaggle/NLP/test.csv"

train = pd.read_csv(train_path)
test = pd.read_csv(test_path)

print("読み込み完了")
print("train shape:", train.shape)
print("test shape :", test.shape)


# =========================
# 2. 必要カラムの確認
# =========================

required_train_cols = ["id", "keyword", "location", "text", "target"]
required_test_cols = ["id", "keyword", "location", "text"]

missing_train_cols = [col for col in required_train_cols if col not in train.columns]
missing_test_cols = [col for col in required_test_cols if col not in test.columns]

if missing_train_cols:
    raise ValueError(f"train.csvに存在しないカラムがあります: {missing_train_cols}")

if missing_test_cols:
    raise ValueError(f"test.csvに存在しないカラムがあります: {missing_test_cols}")


# =========================
# 3. BERT向けの軽いクリーニング関数
# =========================

def clean_text_for_bert(text):
    """
    BERT向けの軽い前処理。
    TF-IDF向けのような強い正規化は行わない。

    やること:
    - HTMLエスケープを戻す
    - URLエンコードを戻す
    - URLを url に置換
    - メンションを user に置換
    - 連続空白を1つにまとめる
    - 前後の空白を削除する

    やらないこと:
    - stopwords除去
    - stemming
    - lemmatization
    - 強い記号削除
    - 明示的な小文字化
      ※ bert-base-uncased の tokenizer 側で小文字化されるため
    """
    if pd.isna(text):
        return ""

    text = str(text)

    # HTMLエスケープを戻す
    text = html.unescape(text)

    # URLエンコードを戻す
    text = unquote(text)

    # URLを url に置換
    text = re.sub(r"https?://\S+|www\.\S+", " url ", text)

    # メンションを user に置換
    text = re.sub(r"@\w+", " user ", text)

    # 連続する空白・改行・タブを1つにまとめる
    text = re.sub(r"\s+", " ", text)

    # 前後の空白を削除
    text = text.strip()

    return text


# =========================
# 4. train / test に同じ前処理を適用
# =========================

train_bert = train.copy()
test_bert = test.copy()

# keywordの欠損値は空文字で補完
train_bert["keyword"] = train_bert["keyword"].fillna("")
test_bert["keyword"] = test_bert["keyword"].fillna("")

# textも念のため欠損値を空文字で補完
train_bert["text"] = train_bert["text"].fillna("")
test_bert["text"] = test_bert["text"].fillna("")

# locationは今回も使わない
# ただし、元カラムは消さずに残す

# BERT向けに軽くクリーニング
train_bert["keyword_bert"] = train_bert["keyword"].apply(clean_text_for_bert)
test_bert["keyword_bert"] = test_bert["keyword"].apply(clean_text_for_bert)

train_bert["text_bert"] = train_bert["text"].apply(clean_text_for_bert)
test_bert["text_bert"] = test_bert["text"].apply(clean_text_for_bert)

# 確認用の input_text_bert も作る
# 実際の学習では tokenizer に keyword_bert と text_bert をペアで渡す
train_bert["input_text_bert"] = np.where(
    train_bert["keyword_bert"] != "",
    train_bert["keyword_bert"] + " [SEP] " + train_bert["text_bert"],
    train_bert["text_bert"]
)

test_bert["input_text_bert"] = np.where(
    test_bert["keyword_bert"] != "",
    test_bert["keyword_bert"] + " [SEP] " + test_bert["text_bert"],
    test_bert["text_bert"]
)

# targetはint型にしておく
train_bert["target"] = train_bert["target"].astype(int)


# =========================
# 5. 確認
# =========================

print("\n===== train_bert head =====")
display(train_bert.head())

print("\n===== test_bert head =====")
display(test_bert.head())

print("\n===== train_bert missing values =====")
display(train_bert.isnull().sum().sort_values(ascending=False))

print("\n===== test_bert missing values =====")
display(test_bert.isnull().sum().sort_values(ascending=False))

print("\n===== input_text_bert samples: train =====")
display(
    train_bert[
        ["id", "keyword", "keyword_bert", "text", "text_bert", "input_text_bert", "target"]
    ].head(10)
)

print("\n===== input_text_bert samples: test =====")
display(
    test_bert[
        ["id", "keyword", "keyword_bert", "text", "text_bert", "input_text_bert"]
    ].head(10)
)

print("\n===== target distribution =====")
display(train_bert["target"].value_counts())
display(train_bert["target"].value_counts(normalize=True))

print("\nBERT用前処理完了")
print("train_bert shape:", train_bert.shape)
print("test_bert shape :", test_bert.shape)
```

```python
!pip install -q -U transformers accelerate

import os
import random
import numpy as np
import pandas as pd
import torch

from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    accuracy_score,
    f1_score,
    classification_report,
    confusion_matrix
)

from transformers import (
    AutoTokenizer,
    AutoModelForSequenceClassification,
    TrainingArguments,
    Trainer
)

# =========================
# 1. 基本設定
# =========================

os.environ["WANDB_DISABLED"] = "true"

MODEL_NAME = "bert-base-uncased"
MAX_LENGTH = 128
RANDOM_STATE = 42

BATCH_SIZE = 16
EVAL_BATCH_SIZE = 32
EPOCHS = 3
LEARNING_RATE = 2e-5
WEIGHT_DECAY = 0.01

device = "cuda" if torch.cuda.is_available() else "cpu"
print("device:", device)


# =========================
# 2. 乱数固定
# =========================

def set_seed(seed=42):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)

set_seed(RANDOM_STATE)


# =========================
# 3. train / valid 分割
# =========================

train_df, valid_df = train_test_split(
    train_bert,
    test_size=0.2,
    stratify=train_bert["target"],
    random_state=RANDOM_STATE
)

train_df = train_df.reset_index(drop=True)
valid_df = valid_df.reset_index(drop=True)

print("train_df shape:", train_df.shape)
print("valid_df shape:", valid_df.shape)

print("\n===== train target distribution =====")
display(train_df["target"].value_counts(normalize=True))

print("\n===== valid target distribution =====")
display(valid_df["target"].value_counts(normalize=True))


# =========================
# 4. Tokenizer
# =========================

tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)


# =========================
# 5. Dataset定義
# =========================

class BertTextDataset(torch.utils.data.Dataset):
    def __init__(self, df, tokenizer, max_length=128, with_labels=True):
        self.df = df.reset_index(drop=True)
        self.tokenizer = tokenizer
        self.max_length = max_length
        self.with_labels = with_labels

        self.keywords = self.df["keyword_bert"].fillna("").tolist()
        self.texts = self.df["text_bert"].fillna("").tolist()

        if self.with_labels:
            self.labels = self.df["target"].astype(int).tolist()

    def __len__(self):
        return len(self.df)

    def __getitem__(self, idx):
        keyword = self.keywords[idx]
        text = self.texts[idx]

        # keyword と text を sentence pair として tokenizer に渡す
        # 内部的には [CLS] keyword [SEP] text [SEP] のように処理される
        encoding = self.tokenizer(
            keyword,
            text,
            truncation=True,
            padding="max_length",
            max_length=self.max_length,
            return_tensors=None
        )

        item = {
            "input_ids": torch.tensor(encoding["input_ids"], dtype=torch.long),
            "attention_mask": torch.tensor(encoding["attention_mask"], dtype=torch.long)
        }

        if "token_type_ids" in encoding:
            item["token_type_ids"] = torch.tensor(encoding["token_type_ids"], dtype=torch.long)

        if self.with_labels:
            item["labels"] = torch.tensor(self.labels[idx], dtype=torch.long)

        return item


train_dataset = BertTextDataset(
    train_df,
    tokenizer=tokenizer,
    max_length=MAX_LENGTH,
    with_labels=True
)

valid_dataset = BertTextDataset(
    valid_df,
    tokenizer=tokenizer,
    max_length=MAX_LENGTH,
    with_labels=True
)


# =========================
# 6. モデル定義
# =========================

model = AutoModelForSequenceClassification.from_pretrained(
    MODEL_NAME,
    num_labels=2
)

model.to(device)


# =========================
# 7. 評価指標
# =========================

def compute_metrics(eval_pred):
    logits, labels = eval_pred
    preds = np.argmax(logits, axis=1)

    acc = accuracy_score(labels, preds)
    f1 = f1_score(labels, preds)

    return {
        "accuracy": acc,
        "f1": f1
    }


# =========================
# 8. TrainingArguments
# =========================

training_args = TrainingArguments(
    output_dir="./bert_base_uncased_results",
    eval_strategy="epoch",
    save_strategy="epoch",
    logging_strategy="steps",
    logging_steps=50,

    learning_rate=LEARNING_RATE,
    per_device_train_batch_size=BATCH_SIZE,
    per_device_eval_batch_size=EVAL_BATCH_SIZE,
    num_train_epochs=EPOCHS,
    weight_decay=WEIGHT_DECAY,

    load_best_model_at_end=True,
    metric_for_best_model="f1",
    greater_is_better=True,
    save_total_limit=2,

    report_to="none",
    fp16=torch.cuda.is_available()
)


# =========================
# 9. Trainer
# =========================

trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=train_dataset,
    eval_dataset=valid_dataset,
    processing_class=tokenizer,
    compute_metrics=compute_metrics
)


# =========================
# 10. 学習
# =========================

trainer.train()

print("\n学習完了")


# =========================
# 11. 検証データで評価
# =========================

eval_results = trainer.evaluate()

print("\n===== eval results =====")
print(eval_results)


# =========================
# 12. classification_report / confusion_matrix
# =========================

valid_output = trainer.predict(valid_dataset)
valid_logits = valid_output.predictions
valid_pred = np.argmax(valid_logits, axis=1)
valid_true = valid_df["target"].values

print("\n===== validation score =====")
print("accuracy:", accuracy_score(valid_true, valid_pred))
print("f1_score:", f1_score(valid_true, valid_pred))

print("\n===== classification report =====")
print(classification_report(valid_true, valid_pred, digits=4))

print("\n===== confusion matrix =====")
cm = confusion_matrix(valid_true, valid_pred)
cm_df = pd.DataFrame(
    cm,
    index=["actual_0", "actual_1"],
    columns=["pred_0", "pred_1"]
)
display(cm_df)


# =========================
# 13. モデル保存
# =========================

save_dir = "./bert_base_uncased_model"

trainer.save_model(save_dir)
tokenizer.save_pretrained(save_dir)

print("\nモデルを保存しました:", save_dir)
```
